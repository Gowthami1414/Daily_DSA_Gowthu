```python
#optimize using two pointer for string palin
def word(s):
  left=0
  righ=len(s)-1
  while left<=len(s)-1:
    if s[left]==s[righ]:
      return 'palin'
    else:
      return 'not palin'
    left+=1
    righ-=1
print(word('madam'))
```

```python
#2.1
def is_palindrome(n):
    original = n
    reversed_num = 0

    while n > 0:
        digit = n % 10
        reversed_num = reversed_num * 10 + digit
        n = n // 10

    return original == reversed_num
print("Palindrome" if is_palindrome(121) else "Not Palindrome")
```

```python
#2.2 p\optimize
def is_palindrome(x):
    # Negative numbers or numbers ending with 0 (but not 0 itself) can't be palindromes
    if x < 0 or (x % 10 == 0 and x != 0):
        return False

    reversed_half = 0
    while x > reversed_half:
        reversed_half = reversed_half * 10 + x % 10
        x //= 10

    # For even-length: x == reversed_half
    # For odd-length: x == reversed_half // 10
    return x == reversed_half or x == reversed_half // 10

# Example usage
print("Palindrome" if is_palindrome(121) else "Not Palindrome")
```

```python
def word(s):
    left = 0
    right = len(s) - 1
    v, c, spa = 0, 0, 0

    while left <= right:
        for ch in (s[left], s[right]):
            if ch in "aeiou":
                v += 1
            elif ch == " ":
                spa += 1
            else:
                c += 1
        left += 1
        right -= 1

    return (v, c, spa)

print(word("gowthami"))
```
