```python
#2.1
def is_palindrome(n):
    original = n
    reversed_num = 0

    while n > 0:
        digit = n % 10
        reversed_num = reversed_num * 10 + digit
        n = n // 10

    return original == reversed_num
print("Palindrome" if is_palindrome(121) else "Not Palindrome")
```

```python
#2.2 p\optimize
def is_palindrome(x):
    # Negative numbers or numbers ending with 0 (but not 0 itself) can't be palindromes
    if x < 0 or (x % 10 == 0 and x != 0):
        return False

    reversed_half = 0
    while x > reversed_half:
        reversed_half = reversed_half * 10 + x % 10
        x //= 10

    # For even-length: x == reversed_half
    # For odd-length: x == reversed_half // 10
    return x == reversed_half or x == reversed_half // 10

# Example usage
print("Palindrome" if is_palindrome(121) else "Not Palindrome")
```

#3.) https://leetcode.com/problems/best-time-to-buy-and-sell-stock/

```python
#3.1#O(N^2)
def buysell(arrbs):
  max_P=0
  for i in range(len(arrbs)):      #n
    for j in range(i+1,len(arrbs)): #n*n
      totbs=arrbs[j]-arrbs[i]
      curr=max(totbs,max_P)
      max_P=curr
  return max_P
print(buysell([7,1,9,3,6,7,4]))
```

```python
#3.2 buysell O(N) #brute force
def buy_sell(nums):
  min_price=nums[0]
  max_price=0
  for price in nums:
    if price<min_price:
      min_price=price
    else:
      max_price=max(max_price,price-min_price)
  return max_price
print(buy_sell([7,1,9,3,6,7,4]))
```

# 4 two sum problem

```python
#4.1

def twoSum(nums, target):                               # TC=O(N)
    hashmap = {}
    for i, num in enumerate(nums):
        complement = target - num
        if complement in hashmap:
            return [hashmap[complement], i]
        hashmap[num] = i
print(twoSum([7,1,9,3,6,7,4],9))
```

```python
#4.2
def twosum(nums, target):               #O(N^2)
  for i in range(len(nums)):
    for j in range(i+1,len(nums)):
      if nums[i]+nums[j]==target:
        return[i,j]
  return[]
twosum([7,1,5,4,2],9)
```

6 .You are given an integer array prices where prices[i] is the price of a given stock on the ith day.

On each day, you may decide to buy and/or sell the stock. You can only hold at most one share of the stock at any time. However, you can buy it then immediately sell it on the same day.

Find and return the maximum profit you can achieve.

```python
def maxProfit(prices):
    profit = 0
    for i in range(1, len(prices)):
        if prices[i] > prices[i - 1]:
            profit += prices[i] - prices[i - 1]
    return profit
```
